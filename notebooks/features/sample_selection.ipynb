```python
import pandas as pd
import numpy as np
from scipy.stats import entropy

df = pd.read_csv("../../data/processed/DogFeatures.csv")
LABEL = "label"
DOGCOL = "DogID"

counts = df.groupby([DOGCOL, LABEL]).size().reset_index(name="count")
counts["total"] = counts.groupby(DOGCOL)["count"].transform("sum")
counts["prop"] = counts["count"] / counts["total"]

def ent(p):
    return entropy(p) / np.log(2)

balance = counts.groupby(DOGCOL).agg(
    entropy=("prop", lambda s: ent(s.values)),
    prop_std=("prop", "std"),
    min_prop=("prop", "min"),
    total_samples=("total", "max")
).reset_index()

balance_filtered = balance[balance["total_samples"] >= 50]
top_by_entropy = balance_filtered.sort_values("entropy", ascending=False).head(10)
top_by_entropy
```
